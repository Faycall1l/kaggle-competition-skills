```python
from pathlib import Path

roots = list(Path("/kaggle/input").rglob("train.csv"))
print("train.csv candidates:", [str(p) for p in roots])
train_path = roots[0]
test_path = next(Path("/kaggle/input").rglob("test.csv"))
```

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression

train = pd.read_csv(train_path)
test = pd.read_csv(test_path)
features = ["Pclass", "Sex", "Age", "SibSp", "Parch", "Fare"]
train[features] = train[features].copy()
train["Sex"] = (train["Sex"] == "female").astype(int)
train["Age"] = train["Age"].fillna(train["Age"].median())
test["Sex"] = (test["Sex"] == "female").astype(int)
test["Age"] = test["Age"].fillna(train["Age"].median())
test["Fare"] = test["Fare"].fillna(test["Fare"].median())

model = LogisticRegression(max_iter=200)
model.fit(train[features].fillna(0), train["Survived"])
preds = model.predict(test[features].fillna(0))
print("train accuracy:", model.score(train[features].fillna(0), train["Survived"]))
```

```python
out = pd.DataFrame({"PassengerId": test["PassengerId"], "Survived": preds})
out.to_csv("/kaggle/working/submission.csv", index=False)
print("wrote", len(out), "rows to /kaggle/working/submission.csv")
```
